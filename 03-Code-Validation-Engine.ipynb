```python
import pandas as pd
import numpy as np

def run_fare_audit(data_path):
    #Loading SQL-processed data
    df = pd.read_csv(data_path)

    #Define booking class rules:
    df['booking_class_violation'] = False

    bk_cl_rules = {
            'F': {'min_days': 0, 'max_days': 7, 'cabin': 'Economy'},
            'H': {'min_days': 8, 'max_days': 42,'cabin': 'Economy'},
            'K': {'min_days': 43, 'max_days': 999, 'cabin': 'Economy'},
            'M': {'min_days': 0, 'max_days': 7, 'cabin': 'Business'},
            'Q': {'min_days': 8, 'max_days': 42, 'cabin': 'Business'},
            'T': {'min_days': 43, 'max_days': 999, 'cabin': 'Business'}
    }

    for bk_cl, rule in bk_cl_rules.items():
        mask = (
            (df['sales_channel'].isin(['GDS'])) &
            (df['booking_class'] == bk_cl) &
            (df['class'] == rule['cabin']) &
            ((df['days_left'] < rule['min_days']) | (df['days_left'] > rule['max_days']))
        )

        df.loc[mask, 'booking_class_violation'] = True

    df.to_csv('vistara_audit_tableau.csv', index=True)
    print("Data saved!")

if __name__ == "__main__":
    audit_df = run_fare_audit('vistara_audit.csv')
```
